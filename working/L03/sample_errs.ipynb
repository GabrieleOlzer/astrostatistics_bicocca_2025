Exercise on sample statistics and the associated errors

```python
import numpy as np
import random
```

```python
n = 10000 #number of data points extracted per iteration
N = 1000 #number of iterations

means = []
std_devs = []

for i in range(N):
    
    data = [random.gauss(3.2, 1.7) for i in range(n+1)]

    means.append(np.mean(data))
    std_devs.append(np.std(data))

#Errors computed as spread of values
means_err_num = np.std(means) 
std_err_num = np.std(std_devs)

#Errors computed using the associted formulae
means_err_th = std_devs[random.randint(0, N-1)]/np.sqrt(n)
std_err_th = std_devs[random.randint(0, N-1)]/np.sqrt(2*(n-1))

print(str(means_err_num) + "   " + str(means_err_th))
print(str(std_err_num) + "   " + str(std_err_th))
```
