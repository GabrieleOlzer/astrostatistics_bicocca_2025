Lecture 02: Monty Hall problem

Numerical simulation of the Monty Hall problem, for each iteration we consider the cases in which a player switches door,
doesn't switch door, and the case of an external observer who sees one door open without knowing the original choice

```python
import numpy as np
import matplotlib as plt
from mpl_toolkits.mplot3d import Axes3D
import random
import pylab
```

```python

n = 10000 #numer of iterations
switch = 0
non_switch = 0
extern = 0

for i in range(n):
    in_choice = random.randint(1, 3)
    
    if in_choice == 1:
        switch_choice = 2
    elif in_choice == 2:
        switch_choice = 1
    elif in_choice == 3:
        switch_choice = 1
    extern_choice = random.randint(1, 2)

    if in_choice == 1:
        non_switch = non_switch + 1
    if switch_choice == 1:
        switch = switch + 1
    if extern_choice == 1:
        extern = extern +1

print("The success probability in the non-switching case is: " + str(non_switch/n))
print("The success probability in the switching case is: " + str(switch/n))
print("The success probability for the external player is: " + str(extern/n))

```

We now generalize the problem to N doors, with the presenter opening p < N of them

```python
N_max = 10
n = 1000000

P_switch = np.zeros((N_max-2, N_max-2), dtype=float)
P_non_switch = np.zeros((N_max-2, N_max-2), dtype=float)
P_extern = np.zeros((N_max-2, N_max-2), dtype=float)

for N in range(3, N_max+1):
    for p in range (1, N-1):

        switch = 0
        non_switch = 0
        extern = 0

        for i in range(n):
            in_choice = random.randint(1, N)

            if in_choice == 1:
                switch_choice = 2
            else:
                a = N-p-1
                switch_choice = random.randint(1, a)
            
            extern_choice = random.randint(1, N-p)

            if in_choice == 1:
                non_switch = non_switch + 1
            if switch_choice == 1:
                switch = switch + 1
            if extern_choice == 1:
                extern = extern +1
        
        P_switch[N-3][p-1] = switch/n
        P_non_switch[N-3][p-1] = non_switch/n
        P_extern[N-3][p-1] = extern/n
            
            
print(P_switch)
print(P_non_switch)
print(P_extern)


```
